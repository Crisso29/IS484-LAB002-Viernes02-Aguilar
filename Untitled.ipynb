## 4.2 Representación del problema mediante estados

### Ficha del problema (El repartidor)

| Campo | Descripción |
|---|---|
| **Estado inicial** | Almacén |
| **Estado objetivo** | ZonaNorte |
| **Acciones posibles** | Moverse a un punto directamente conectado |
| **Costo de cada acción** | 1 paso por movimiento (costo uniforme) |
| **Espacio de estados** | {Almacén, Centro, Terminal, Mercado, ZonaNorte} |
| **Solución** | Secuencia de puntos desde Almacén hasta ZonaNorte |


````python
# Representación del mapa como grafo (diccionario de adyacencia)
mapa = {
    "Almacen":   ["Centro", "Terminal"],
    "Centro":    ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal":  ["Almacen", "Mercado"],
    "Mercado":   ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print(mapa)
````

## 4.3 Implementación de BFS (Búsqueda en Amplitud)

````python
def bfs(grafo, inicio, objetivo):
    cola = [[inicio]]   # cola de caminos
    visitados = set()

    while cola:
        camino = cola.pop(0)        # saco el primero (FIFO)
        nodo_actual = camino[-1]    # el último nodo del camino actual

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # extiendo el camino
                cola.append(nuevo_camino)         # lo meto al final de la cola

    return None

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (BFS):", resultado)
````

###  Traza de ejecución_BFS (FIFO)

| Iteración | Saco de la cola | Meto a la cola |
|:---------:|----------------|----------------|
| **Inicio** | — | `["Almacen"]` |
| **1** | `["Almacen"]` | `["Almacen", "Centro"]` <br> `["Almacen", "Terminal"]` |
| **2** | `["Almacen", "Centro"]` | `["Almacen", "Centro", "ZonaNorte"]` <br> `["Almacen", "Centro", "Mercado"]` |
| **3** | `["Almacen", "Terminal"]` | `["Almacen", "Terminal", "Mercado"]` |
| **4** | `["Almacen", "Centro", "ZonaNorte"]` | — |

>  `nodo_actual == "ZonaNorte"` → se retorna el camino encontrado

```python
["Almacen", "Centro", "ZonaNorte"]
```
>  **BFS garantiza el camino con menos pasos** porque siempre explora nivel por nivel antes de ir más profundo.
