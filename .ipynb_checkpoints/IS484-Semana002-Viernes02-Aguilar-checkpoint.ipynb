## 4.2 Representación del problema mediante estados

### Ficha del problema (El repartidor)

| Campo | Descripción |
|---|---|
| **Estado inicial** | Almacén |
| **Estado objetivo** | ZonaNorte |
| **Acciones posibles** | Moverse a un punto directamente conectado |
| **Costo de cada acción** | 1 paso por movimiento (costo uniforme) |
| **Espacio de estados** | {Almacén, Centro, Terminal, Mercado, ZonaNorte} |
| **Solución** | Secuencia de puntos desde Almacén hasta ZonaNorte |


````python
# Representación del mapa como grafo (diccionario de adyacencia)
mapa = {
    "Almacen":   ["Centro", "Terminal"],
    "Centro":    ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal":  ["Almacen", "Mercado"],
    "Mercado":   ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}

print(mapa)
````

## 4.3 Implementación de BFS (Búsqueda en Amplitud)

````python
def bfs(grafo, inicio, objetivo):
    cola = [[inicio]]   # cola de caminos
    visitados = set()

    while cola:
        camino = cola.pop(0)        # saco el primero (FIFO)
        nodo_actual = camino[-1]    # el último nodo del camino actual

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # extiendo el camino
                cola.append(nuevo_camino)         # lo meto al final de la cola

    return None

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (BFS):", resultado)
````

###  Traza de ejecución_BFS (FIFO)

| Iteración | Saco de la cola | Meto a la cola |
|:---------:|----------------|----------------|
| **Inicio** | — | `["Almacen"]` |
| **1** | `["Almacen"]` | `["Almacen", "Centro"]` <br> `["Almacen", "Terminal"]` |
| **2** | `["Almacen", "Centro"]` | `["Almacen", "Centro", "ZonaNorte"]` <br> `["Almacen", "Centro", "Mercado"]` |
| **3** | `["Almacen", "Terminal"]` | `["Almacen", "Terminal", "Mercado"]` |
| **4** | `["Almacen", "Centro", "ZonaNorte"]` | — |

>  `nodo_actual == "ZonaNorte"` → se retorna el camino encontrado

```python
["Almacen", "Centro", "ZonaNorte"]
```
>  **BFS garantiza el camino con menos pasos** porque siempre explora nivel por nivel antes de ir más profundo.

## 4.4 Vectores y Matrices con NumPy

````python
import numpy as np
obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)
````

````python
# Acceder a filas y columnas
print("Primera observación (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
````

## 4.5. Producto escalar y distancia euclidiana

**Parte A:** Calcular el producto escalar entre dos observaciones:""

````python
producto = np.dot(obs1, obs2)
print("Producto escalar entre obs1 y obs2:", producto)
````

**Parte B**: Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual
y luego con NumPy:

````python
# Formula manual
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)
# Con la funcion de NumPy
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
````

## 5.2 Implementación de DFS (Búsqueda en Profundidad)

El DFS usa una **pila LIFO** (Last In, First Out): el último camino en entrar es el primero en salir.  
A diferencia del BFS que explora nivel por nivel, el DFS se **hunde por una rama** hasta el final antes de retroceder.

````python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    Usa una pila LIFO: .pop() saca el ÚLTIMO elemento.
    """
    pila      = [[inicio]]  # pila de caminos
    visitados = set()

    while pila:
        camino      = pila.pop()     # saco el ÚLTIMO (LIFO)
        nodo_actual = camino[-1]     # nodo donde estamos parados

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]  # extiendo el camino
                pila.append(nuevo_camino)         # meto al final de la pila

    return None

resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
````

### 🔍 Traza de ejecución — DFS (LIFO)

| Iteración | Saco de la pila | Meto a la pila |
|:---------:|----------------|----------------|
| **Inicio** | — | `["Almacen"]` |
| **1** | `["Almacen"]` | `["Almacen","Centro"]` <br> `["Almacen","Terminal"]` |
| **2** | `["Almacen","Terminal"]` ← *el último* | `["Almacen","Terminal","Almacen"]` <br> `["Almacen","Terminal","Mercado"]` |
| **3** | `["Almacen","Terminal","Mercado"]` ← *el último* | `["Almacen","Terminal","Mercado","Centro"]` <br> `["Almacen","Terminal","Mercado","Terminal"]` <br> `["Almacen","Terminal","Mercado","ZonaNorte"]` |
| **4** | `["Almacen","Terminal","Mercado","ZonaNorte"]` ← *el último* | — |

> ✅ `nodo_actual == "ZonaNorte"` → se retorna el camino encontrado

### 📊 Comparación BFS vs DFS

| | BFS (FIFO) | DFS (LIFO) |
|---|---|---|
| **Estructura** | Cola — saca el primero | Pila — saca el último |
| **Estrategia** | Nivel por nivel | Se hunde por una rama |
| **Camino encontrado** | `Almacen → Centro → ZonaNorte` | `Almacen → Terminal → Mercado → ZonaNorte` |
| **Pasos** | ✅ 2 pasos | ❌ 3 pasos |
| **¿Garantiza el más corto?** | ✅ Sí (costo uniforme) | ❌ No |

> 📌 **Conclusión:** BFS encontró el camino óptimo en 2 pasos.  
> DFS llegó al objetivo también, pero tomó 3 pasos porque se hundió  
> por la rama `Terminal → Mercado` antes de descubrir el atajo por `Centro`.

## 5.3 Encontrar la Observación más Cercana

````python
nueva_observacion = np.array([20.5, 48])

distancias = []

for i, obs in enumerate(dataset):
    d = np.linalg.norm(nueva_observacion - obs)
    distancias.append(d)
    print(f"Distancia a la observación {i}: {d:.4f}")

indice_mas_cercano = np.argmin(distancias)
print("\nLa observación más parecida es la número", indice_mas_cercano)
print("Sus valores son:", dataset[indice_mas_cercano])
````

## 5.4 Reflexión

**¿Por qué la distancia euclidiana es una forma razonable de medir 
qué tan parecidas son dos observaciones?**

Porque mide geométricamente qué tan lejos está un punto de otro 
en el espacio de características. Si dos observaciones tienen valores 
similares en todas sus dimensiones, su distancia será pequeña, 
lo que indica que son parecidas. Es la generalización del teorema 
de Pitágoras a n dimensiones.

**¿En qué casos podría no ser la mejor opción?**

Cuando las características tienen escalas muy distintas. Por ejemplo, 
si combinamos temperatura (15–35 °C) con ingreso mensual (800–8000 soles), 
la distancia quedará dominada por la variable de mayor escala (el ingreso), 
haciendo que la temperatura casi no influya en el resultado. 
La solución es **normalizar** los datos antes de calcular distancias, 
técnica que se verá más adelante en el curso.
